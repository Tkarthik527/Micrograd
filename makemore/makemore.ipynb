```python
data = open("names.txt").read().splitlines()
```

```python
data[:20]
```

```python
map = {}
```

```python
alpha =[ chr(i) for i in range(ord('a')-1, ord('z')+1)]
```

```python
map = { chr(i) : alpha.index(chr(i))  for i in range(ord('a')-1, ord('z')+1)}
```

```python
map['.'] = 0
```

```python
import torch

dataset = torch.zeros(27,27 , dtype=torch.int32)
```

```python
for word in data:
    word = '.' + word + '.'
    for chr1 , chr2 in zip(word, word[1:]):
        dataset[ map[chr1] ,  map[chr2] ] += 1    
```

```python
from matplotlib import pyplot

pyplot.figure(figsize=(5,5))
pyplot.imshow(dataset, cmap='Blues')
```
